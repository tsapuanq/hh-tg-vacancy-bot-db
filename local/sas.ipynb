```python
import pandas as pd
```

```python
df = pd.read_csv('analytics_check.csv')
df.info()
```

```python
titles = df['title'].fillna('').unique().tolist()
print(titles[:50])  # первые 50 уникальных заголовков
```

```python
import datetime 
today = datetime.datetime.today()
week = today - datetime.timedelta(days=7)
df['published_at'] = pd.to_datetime(df['published_at'])
last_week_vacancies = df[df['published_at'] >= week]
last_week_vacancies
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
df['published_at'] = pd.to_datetime(df['published_at'])
df['day'] = df['published_at'].dt.date
df['month'] = df['published_at'].dt.to_period('M')
plt.figure(figsize=(12,5))
sns.countplot(x='day', data=df, order=sorted(df['day'].unique()))
plt.xticks(rotation=45)
plt.title('Количество вакансий по дням')
plt.show()
plt.figure(figsize=(10,5))
sns.countplot(x='month', data=df, order=sorted(df['month'].unique()))
plt.title('Количество вакансий по месяцам')
plt.show()
```

```python
df['month'] = df['published_at'].dt.to_period('M')
df['month']
```

```python
# Визуализация количества вакансий по компаниям за месяц
if 'company' in df.columns:
    company_month = df.groupby(['company', 'month']).size().reset_index(name='vacancy_count')
    plt.figure(figsize=(14,6))
    sns.barplot(x='month', y='vacancy_count', hue='company', data=company_month)
    plt.title('Количество вакансий по компаниям за месяц')
    plt.xticks(rotation=45)
    plt.legend(title='Компания', bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.show()
else:
    print('В таблице нет столбца company')
```

```python
# Визуализация: компании с количеством вакансий за месяц больше 3
if 'company' in df.columns:
    company_month_table = df.groupby(['company', 'month']).size().reset_index(name='vacancy_count')
    filtered_table = company_month_table[company_month_table['vacancy_count'] > 3]
    plt.figure(figsize=(14,6))
    sns.barplot(x='month', y='vacancy_count', hue='company', data=filtered_table)
    plt.title('Компании с количеством вакансий > 3 за месяц')
    plt.xticks(rotation=45)
    plt.legend(title='Компания', bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.show()
else:
    print('В таблице нет столбца company')
```

```python
# Визуализация: отдельные графики для топ-5 компаний по количеству вакансий за месяц
if 'company' in df.columns:
    company_month_table = df.groupby(['company', 'month']).size().reset_index(name='vacancy_count')
    top_companies = company_month_table.groupby('company')['vacancy_count'].sum().nlargest(5).index
    for company in top_companies:
        data = company_month_table[company_month_table['company'] == company]
        plt.figure(figsize=(8,4))
        sns.barplot(x='month', y='vacancy_count', data=data)
        plt.title(f'Вакансии по месяцам: {company}')
        plt.xticks(rotation=45)
        plt.ylabel('Количество вакансий')
        plt.xlabel('Месяц')
        plt.tight_layout()
        plt.show()
else:
    print('В таблице нет столбца company')
```
